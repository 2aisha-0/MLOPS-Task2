```python
import pandas as pd

import os



# ============================================================

# 1. CHECK DATA DIRECTORY

# ============================================================



print("DATA FILES:")

print(os.listdir("/app/data"))





# ============================================================

# 2. READ ALL DATASETS

# ============================================================



customers = pd.read_csv("/app/data/olist_customers_dataset.csv")

geolocation = pd.read_csv("/app/data/olist_geolocation_dataset.csv")

orders = pd.read_csv("/app/data/olist_orders_dataset.csv")

order_items = pd.read_csv("/app/data/olist_order_items_dataset.csv")

order_payments = pd.read_csv("/app/data/olist_order_payments_dataset.csv")

order_reviews = pd.read_csv("/app/data/olist_order_reviews_dataset.csv")

products = pd.read_csv("/app/data/olist_products_dataset.csv")

sellers = pd.read_csv("/app/data/olist_sellers_dataset.csv")

category_translation = pd.read_csv(

"/app/data/product_category_name_translation.csv"

)





# ============================================================

# 3. CUSTOMERS

# ============================================================



print("\n========== CUSTOMERS ==========")



print("Shape:", customers.shape)

print("Columns:")

print(customers.columns)



print("Number of rows:", len(customers))

print("Unique customer_id:", customers["customer_id"].nunique())

print(

"Is customer_id unique?",

len(customers) == customers["customer_id"].nunique()

)



print("\nUnique customer_unique_id:")

print(customers["customer_unique_id"].nunique())



print("\nCustomer_id duplicated:")

print(customers["customer_id"].duplicated().sum())





# ============================================================

# 4. ORDERS

# ============================================================



print("\n========== ORDERS ==========")



print("Shape:", orders.shape)

print("Columns:")

print(orders.columns)



print("Number of rows:", len(orders))

print("Unique order_id:", orders["order_id"].nunique())



print(

"Is order_id unique?",

len(orders) == orders["order_id"].nunique()

)



print("\nOrder status distribution:")

print(orders["order_status"].value_counts())





# ============================================================

# 5. ORDER ITEMS

# ============================================================



print("\n========== ORDER ITEMS ==========")



print("Shape:", order_items.shape)

print("Columns:")

print(order_items.columns)



print("Number of rows:", len(order_items))

print("Unique order_id:", order_items["order_id"].nunique())

print("Unique product_id:", order_items["product_id"].nunique())

print("Unique seller_id:", order_items["seller_id"].nunique())



print("\nNumber of items per order:")

print(

order_items.groupby("order_id")

.size()

.value_counts()

.sort_index()

)



print("\nNumber of items per order (summary):")

print(

order_items.groupby("order_id")

.size()

.describe()

)





# ============================================================

# 6. ORDER PAYMENTS

# ============================================================



print("\n========== ORDER PAYMENTS ==========")



print("Shape:", order_payments.shape)

print("Columns:")

print(order_payments.columns)



print("Number of rows:", len(order_payments))

print("Unique order_id:", order_payments["order_id"].nunique())



print("\nNumber of payments per order:")

print(

order_payments.groupby("order_id")

.size()

.value_counts()

.sort_index()

)



print("\nPayment type distribution:")

print(order_payments["payment_type"].value_counts())





# ============================================================

# 7. ORDER REVIEWS

# ============================================================



print("\n========== ORDER REVIEWS ==========")



print("Shape:", order_reviews.shape)

print("Columns:")

print(order_reviews.columns)



print("Number of rows:", len(order_reviews))

print("Unique order_id:", order_reviews["order_id"].nunique())



print("\nNumber of reviews per order:")

print(

order_reviews.groupby("order_id")

.size()

.value_counts()

.sort_index()

)





# ============================================================

# 8. PRODUCTS

# ============================================================



print("\n========== PRODUCTS ==========")



print("Shape:", products.shape)

print("Columns:")

print(products.columns)



print("Number of rows:", len(products))

print("Unique product_id:", products["product_id"].nunique())



print(

"Is product_id unique?",

len(products) == products["product_id"].nunique()

)



print("\nMissing product category:")

print(products["product_category_name"].isna().value_counts())





# ============================================================

# 9. SELLERS

# ============================================================



print("\n========== SELLERS ==========")



print("Shape:", sellers.shape)

print("Columns:")

print(sellers.columns)



print("Number of rows:", len(sellers))

print("Unique seller_id:", sellers["seller_id"].nunique())



print(

"Is seller_id unique?",

len(sellers) == sellers["seller_id"].nunique()

)





# ============================================================

# 10. GEOLOCATION

# ============================================================



print("\n========== GEOLOCATION ==========")



print("Shape:", geolocation.shape)

print("Columns:")

print(geolocation.columns)



print("Number of rows:", len(geolocation))



print(

"Unique geolocation_zip_code_prefix:",

geolocation["geolocation_zip_code_prefix"].nunique()

)



print(

"Duplicated zip prefixes:",

geolocation["geolocation_zip_code_prefix"].duplicated().sum()

)





# ============================================================

# 11. CATEGORY TRANSLATION

# ============================================================



print("\n========== CATEGORY TRANSLATION ==========")



print("Shape:", category_translation.shape)

print("Columns:")

print(category_translation.columns)



print("Number of rows:", len(category_translation))



print(

"Unique product categories:",

category_translation["product_category_name"].nunique()

)



print(

"Is product_category_name unique?",

len(category_translation)

== category_translation["product_category_name"].nunique()

)





# ============================================================

# 12. CHECK RELATIONSHIPS

# ============================================================



print("\n========== RELATIONSHIPS ==========")



print("\nORDERS -> CUSTOMERS")



print(

"Unique customer_id in orders:",

orders["customer_id"].nunique()

)



print(

"Unique customer_id in customers:",

customers["customer_id"].nunique()

)





print("\nORDERS -> ORDER ITEMS")



print(

"Orders:",

orders["order_id"].nunique()

)



print(

"Orders appearing in order_items:",

order_items["order_id"].nunique()

)





print("\nORDER ITEMS -> PRODUCTS")



print(

"Products:",

products["product_id"].nunique()

)



print(

"Products appearing in order_items:",

order_items["product_id"].nunique()

)





print("\nORDER ITEMS -> SELLERS")



print(

"Sellers:",

sellers["seller_id"].nunique()

)



print(

"Sellers appearing in order_items:",

order_items["seller_id"].nunique()

)





print("\nORDERS -> PAYMENTS")



print(

"Orders with payments:",

order_payments["order_id"].nunique()

)





print("\nORDERS -> REVIEWS")



print(

"Orders with reviews:",

order_reviews["order_id"].nunique()

)





# ============================================================

# 13. IMPORTANT: CHECK THE GRAIN

# ============================================================



print("\n========== GRAIN CHECK ==========")



print(

"orders rows:",

len(orders)

)



print(

"unique order_id:",

orders["order_id"].nunique()

)



print(

"1 row = 1 order:",

len(orders) == orders["order_id"].nunique()

)





# ============================================================

# 14. SUMMARY

# ============================================================



print("\n========== SUMMARY ==========")



print("""

Base table:

orders



Target grain:

1 row = 1 order



One-to-many relationships:

orders -> order_items

orders -> order_payments

orders -> order_reviews



Many-to-one relationships:

order_items -> products

order_items -> sellers



Customer relationship:

orders -> customers



Product category relationship:

products -> category_translation



Geolocation:

customer/seller ZIP prefix -> geolocation

but geolocation_zip_code_prefix is NOT unique.

""")  



```

```python
# ============================================================
# JOINS + AGGREGATIONS
# Target grain: 1 row = 1 order
# ============================================================

# 1. ORDERS + CUSTOMERS
orders_customers = orders.merge(
    customers,
    on="customer_id",
    how="left"
)


# 2. ORDER ITEMS
items_agg = order_items.groupby("order_id").agg(
    item_count=("order_item_id", "count"),
    unique_products=("product_id", "nunique"),
    unique_sellers=("seller_id", "nunique"),
    total_price=("price", "sum"),
    total_freight=("freight_value", "sum")
).reset_index()

orders_customers_items = orders_customers.merge(
    items_agg,
    on="order_id",
    how="left"
)


# 3. ORDER PAYMENTS
payments_agg = order_payments.groupby("order_id").agg(
    payment_count=("payment_sequential", "count"),
    total_payment=("payment_value", "sum"),
    payment_types=("payment_type", lambda x: ", ".join(x.unique()))
).reset_index()

orders_customers_items_payments = orders_customers_items.merge(
    payments_agg,
    on="order_id",
    how="left"
)


# 4. ORDER REVIEWS
reviews_agg = order_reviews.groupby("order_id").agg(
    review_count=("review_id", "count"),
    average_review_score=("review_score", "mean")
).reset_index()

final_orders = orders_customers_items_payments.merge(
    reviews_agg,
    on="order_id",
    how="left"
)


# ============================================================
# CHECK FINAL GRAIN
# ============================================================

print("Final shape:", final_orders.shape)
print("Rows:", len(final_orders))
print("Unique orders:", final_orders["order_id"].nunique())
print(
    "1 row = 1 order:",
    len(final_orders) == final_orders["order_id"].nunique()
)

print("\nFinal columns:")
print(final_orders.columns.tolist())

print("\nPreview:")
print(final_orders.head())
```

```python
# ============================================================
# 5. PRODUCTS
# ============================================================

items_products = order_items.merge(
    products,
    on="product_id",
    how="left"
)

products_agg = items_products.groupby("order_id").agg(
    product_count=("product_id", "nunique"),
    missing_product_category=("product_category_name", lambda x: x.isna().sum())
).reset_index()

final_orders = final_orders.merge(
    products_agg,
    on="order_id",
    how="left"
)

print("Shape:", final_orders.shape)
print("Unique orders:", final_orders["order_id"].nunique())
print(
    "1 row = 1 order:",
    len(final_orders) == final_orders["order_id"].nunique()
)

print(final_orders[[
    "order_id",
    "product_count",
    "missing_product_category"
]].head())
```

```python
# ============================================================
# 6. SELLERS
# ============================================================

items_sellers = order_items.merge(
    sellers,
    on="seller_id",
    how="left"
)

sellers_agg = items_sellers.groupby("order_id").agg(
    seller_count=("seller_id", "nunique"),
    seller_states=("seller_state", "nunique")
).reset_index()

final_orders = final_orders.merge(
    sellers_agg,
    on="order_id",
    how="left"
)

print("Shape:", final_orders.shape)
print("Unique orders:", final_orders["order_id"].nunique())

print(
    "1 row = 1 order:",
    len(final_orders) == final_orders["order_id"].nunique()
)

print(
    final_orders[
        [
            "order_id",
            "seller_count",
            "seller_states"
        ]
    ].head()
)
```

```python
# ============================================================
# 7. CATEGORY TRANSLATION
# ============================================================

products_categories = products.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

items_categories = order_items.merge(
    products_categories[
        [
            "product_id",
            "product_category_name",
            "product_category_name_english"
        ]
    ],
    on="product_id",
    how="left"
)

categories_agg = items_categories.groupby("order_id").agg(
    category_count=("product_category_name_english", "nunique")
).reset_index()

final_orders = final_orders.merge(
    categories_agg,
    on="order_id",
    how="left"
)

print("Shape:", final_orders.shape)
print("Unique orders:", final_orders["order_id"].nunique())

print(
    "1 row = 1 order:",
    len(final_orders) == final_orders["order_id"].nunique()
)

print(
    final_orders[
        [
            "order_id",
            "category_count"
        ]
    ].head()
)
```

```python
# 8. GEOLOCATION
# ============================================================

geolocation_agg = geolocation.groupby(
    "geolocation_zip_code_prefix"
).agg(
    geo_lat=("geolocation_lat", "mean"),
    geo_lng=("geolocation_lng", "mean"),
    geo_city_count=("geolocation_city", "nunique"),
    geo_state_count=("geolocation_state", "nunique")
).reset_index()
```

```python
final_orders = final_orders.merge(
    geolocation_agg,
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left"
)

final_orders = final_orders.drop(
    columns=["geolocation_zip_code_prefix"]
)
```
