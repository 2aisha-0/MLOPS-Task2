```python
import pandas as pd
import os
```

```python
final_orders = pd.read_csv("/app/data/final_orders.csv")
```

```python
final_orders.shape
```

```python
final_orders["order_delivered_customer_date"]
```

```python
labeled_orders = final_orders[
    final_orders["order_delivered_customer_date"].notna()
].copy()
```

```python
labeled_orders.shape
```

```python
date_columns = [
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

labeled_orders[date_columns] = labeled_orders[date_columns].apply(
    pd.to_datetime
)
```

```python
labeled_orders[date_columns].dtypes
```

```python
labeled_orders["is_late"] = (
    labeled_orders["order_delivered_customer_date"]
    > labeled_orders["order_estimated_delivery_date"]
).astype(int)
```

```python
labeled_orders[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "is_late"
    ]
].head(10)
```

```python
labeled_orders[
    labeled_orders["is_late"] == 1
][
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "is_late"
    ]
].head(10)
```

```python
labeled_orders["is_late"].value_counts()
```

```python
labeled_orders["is_late"].value_counts(normalize=True) * 100
```

```python
labeled_orders["is_late"].isna().sum()
```

```python
print("Rows:", len(labeled_orders))
print("Unique orders:", labeled_orders["order_id"].nunique())
print(
    "1 row = 1 order:",
    len(labeled_orders) == labeled_orders["order_id"].nunique()
)
```

```python
labeled_orders.to_csv(
    "/app/data/labeled_orders.csv",
    index=False
)
```
