```python
import pandas as pd 
import os 
```

```python
labeled_orders=pd.read_csv(
"/app/data/labeled_orders.csv"
)
```

```python
labeled_orders.shape
```

```python
X = labeled_orders.drop(columns=["is_late"])
y = labeled_orders["is_late"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    random_state=42,
    stratify=y
)
```

```python
print("Train + Validation:", X_train_val.shape)
print("Test:", X_test.shape)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.17647,
    random_state=42,
    stratify=y_train_val
)
```

```python
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
print("Train:")
print(y_train.value_counts(normalize=True) * 100)

print("\nValidation:")
print(y_val.value_counts(normalize=True) * 100)

print("\nTest:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
print("Train rows:", len(X_train))
print("Validation rows:", len(X_val))
print("Test rows:", len(X_test))

print("Total rows:", len(X_train) + len(X_val) + len(X_test))
print("Original rows:", len(labeled_orders))
```

```python
print(
    "Train ∩ Validation:",
    len(set(X_train["order_id"]) & set(X_val["order_id"]))
)

print(
    "Train ∩ Test:",
    len(set(X_train["order_id"]) & set(X_test["order_id"]))
)

print(
    "Validation ∩ Test:",
    len(set(X_val["order_id"]) & set(X_test["order_id"]))
)
```

```python
# Save Train / Validation / Test splits

X_train.to_csv("/app/data/X_train.csv", index=False)
X_val.to_csv("/app/data/X_val.csv", index=False)
X_test.to_csv("/app/data/X_test.csv", index=False)

y_train.to_csv("/app/data/y_train.csv", index=False)
y_val.to_csv("/app/data/y_val.csv", index=False)
y_test.to_csv("/app/data/y_test.csv", index=False)

print("Splits saved successfully.")
```
