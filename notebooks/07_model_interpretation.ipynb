```python
# Import pandas for working with the final model interpretation results
import pandas as pd

# Import matplotlib for visualizing model interpretation results
import matplotlib.pyplot as plt
```

```python
# Load the feature importance results produced during model training
feature_importance = pd.read_csv(
    "/app/data/feature_importance.csv"
)

# Sort features from the most important to the least important
feature_importance = feature_importance.sort_values(
    by="importance",
    ascending=False
)

# Display the top 15 most important features
feature_importance.head(15)
```

```python
# Select the top 10 features based on permutation importance
top_10_features = feature_importance.head(10).copy()

# Display the top 10 features and their importance values
top_10_features
```

```python
# Sort the top features for a horizontal bar chart
top_10_plot = top_10_features.sort_values(
    by="importance",
    ascending=True
)

# Create a horizontal bar chart of the top 10 features
plt.figure(figsize=(10, 6))

plt.barh(
    top_10_plot["feature"],
    top_10_plot["importance"]
)

plt.xlabel("Permutation Importance")
plt.ylabel("Feature")
plt.title("Top 10 Features Influencing Model Performance")

plt.tight_layout()
plt.show()
```

```python
# Store the final model and test-set results from the completed evaluation
final_model_summary = pd.DataFrame({
    "Model": ["HistGradientBoostingClassifier"],
    "Threshold": [0.15],
    "Accuracy": [0.857863],
    "Precision": [0.277582],
    "Recall": [0.469336],
    "F1": [0.348845],
    "ROC-AUC": [0.782852]
})

# Display the final model summary
final_model_summary
```
